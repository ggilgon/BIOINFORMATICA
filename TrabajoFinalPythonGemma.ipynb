#### 5.- Devuelve la secuencia de ARN que surgiría a partir de la siguiente secuencia. Recuerda que la secuencia de ARN resultante tiene que ser complementaria a la de ADN. (1 pto)

```python
#defino la funcion para cambiar la secuencia a ARN, cambiando en la secuencia las A por T, las T por U, las C por G, y las G por C. Para que no me cambie todos los elementos de una, hago que lo vaya leyendo uno a uno haciendo una variable nueva, y metiendo elemento por elemento en ella uno a uno con el elif y el +=
def ARN(calculo_CG):
    arnbase = ""
    for i in calculo_CG:
        if i == 'A':
            arnbase += 'T'
        elif i == 'T':
            arnbase += 'U'
        elif i == 'C':
            arnbase += 'G'
        elif i == 'G':
            arnbase += 'C'
    return arnbase
```

```python
secuencia_arn = ARN(calculo_CG)
```

#### 6.- Realiza una función que realice una digestión con las siguientes enzimas de restricción en nuestra secuencia. La función tiene que devolver la siguiente información: (2 pto)
1. Enzimas que tienen sitios de corte en la secuencia y posición (o posiciones de corte son el asterisco)
2. Enzimas que no tienen sitios de corte (si las hay)
3. Resultado de la digestión: los fragmentos resultantes de nuestra secuencia
4. Indicar número y tamaño de cada fragmento

```python
enzR = {
    'HindIII' : 'A*AGCTT',
    'BamHI' : 'G*GATCC',
    'AluI' : 'AG*CT',
    'Sau3AI' : '*GATC',
    'EcoRI' : 'G*AATTC'
}
```

```python
for key,valor in enzR.items(): 
    patron = valor.replace("*","")
    print(patron)
```

```python
#definimos la funcion para que se haga la digestion
#añadimos varios diccionarios para poder ir incluyendo tanto las enzimas como la posicion, y aquellas enzimas que no tienen corte, y los fragmentos tras a digestion.
def digestion(calculo_CG, enzR):
    enzimascorte = {}
    enzimasnocorte = []
    fragmentos = []

    #remplazamos el asterisco para que no nos de ningún error cuando haya que buscar la secuencia que coincide
    for key,valor in enzR.items(): 
        patron = valor.replace("*","")
        corte = valor.index("*")
        
        #guardamos la posicion donde aparece el patron desde el inicio
        posicion = [m.start() + corte for m in re.finditer(patron, calculo_CG)]

        #si la secuencia que corta la enzima coincide con la posicion, hay que añadirla como key, porque despues dentro de esa key se añadira la posicion de corte y los fragmentos
        #si no coincide se añadira al diccionario de enzimasnocorte
        if posicion:
            enzimascorte[key] = posicion
        else:
            enzimasnocorte.append(key)

    #almacenamos los cortes que se producen
    cortes = []
    for posicion in enzimascorte.values():
        cortes.extend(posicion)

    #para agregarlo desde el inicio hasta el final
    no_duplicar_cortes = sorted(set(cortes))
    posicion_corte = [0] + no_duplicar_cortes + [len(calculo_CG)]

    #para tener la secuencia en los fragmentos cortados
    #indica y mete en una lista las posiciones de los cortes, y la longitud de ellos
    #fragmento es , en la secuencia de adn, desde la posicion de corte que se ha obtenido hasta la siguiente posicion de corte.
    for i in range(len(posicion_corte) - 1):
        fragmento = calculo_CG[posicion_corte[i]:posicion_corte[i+1]]
        fragmentos.append(fragmento)

    #para mostrar los resultados
        #si la enzima corta, que me muestre que enzima y done
    for enzima, posicion in enzimascorte.items():
            print(f'{enzima}: corte en {posicion}')

    #si la enzima no corte, que solo me nombre la enzima
    if enzimasnocorte:
        for enzima in enzimasnocorte:
            print(enzima)

    #para que te muestre los fragmentos
    #por cada elemento, recorra la lista y me indice el numero del elemento, es decir donde se encuentra y en la lista hecha y el fragmento en si
    for i, frag in enumerate(fragmentos):
        print(f'Fragmento {i+1}: {len(frag)} pares de bases - {frag}')

    #para que me muestre el numero de fragmentos totales y su tamaño
    print(f'Número total de fragmentos: {len(fragmentos)}')

    return enzimascorte, enzimasnocorte, fragmentos
```

```python
digestion(calculo_CG, enzR)
```

```python
tabla_aa = {
    'UUU':'F','UUC':'F','UUA':'L','UUG':'L','CUU':'L','CUC':'L','CUA':'L','CUG':'L',
        'AUU':'I','AUC':'I','AUA':'I','AUG':'M','GUU':'V','GUC':'V','GUA':'V','GUG':'V',
        'UCU':'S','UCC':'S','UCA':'S','UCG':'S','CCU':'P','CCC':'P','CCA':'P','CCG':'P',
        'ACU':'T','ACC':'T','ACA':'T','ACG':'T','GCU':'A','GCC':'A','GCA':'A','GCG':'A',
        'UAU':'Y','UAC':'Y','UAA':'*','UAG':'*','CAU':'H','CAC':'H','CAA':'Q','CAG':'Q',
        'AAU':'N','AAC':'N','AAA':'K','AAG':'K','GAU':'D','GAC':'D','GAA':'E','GAG':'E',
        'UGU':'C','UGC':'C','UGA':'*','UGG':'W','CGU':'R','CGC':'R','CGA':'R','CGG':'R',
        'AGU':'S','AGC':'S','AGA':'R','AGG':'R','GGU':'G','GGC':'G','GGA':'G','GGG':'G'
    }

#para encontrar los orf, utilizaremos la secuencia de adn, sin embargo para la traduccion a proteina utilizaremos la de arn
#definimos la funcion, y hacemos una lista donde guardaremos los orf encontrados.
#Por cada elemento en la secuencia, en un rago de 3 elementos o nucleotidos, siempre que el elemento sea menor que esa longitud menos 2,
#el codon se va a formar por ese elemento o nucleotido, mas los dos siguientes.
#si el codon es ATG, vamos al siguiente elemento (j es ortro elemento de la cadena para no repetir i)
#se repite el mismo procedimiento
#si el siguiente codon es algunos de los de tipo STOP, se añade ese orf en la lista creada y se para ese bucle.
def orf(calculo_CG):
    orfs = {}
    contador = 0
    for frame in range(3):
        i = frame
        while i < len(calculo_CG) - 2:
            codon = calculo_CG[i:i+3]
            if codon == 'ATG':
                j = i
                while j < len(calculo_CG) - 2:
                    codon2 = calculo_CG[j:j+3]
                    if codon2 in ['TAA', 'TAG', 'TGA']:
                        orf = calculo_CG[i:j+3]
                        orfs[contador] = calculo_CG[i:j+3]
                        #orfs.append([orf, i+1, signo_cadena])
                        contador += 1
                        break
                    j += 3
            i += 3
        return orfs


#tenemos que buscar ORFs en las dos cadenas, por lo que asi ya le damos valor al argumento de la funcion signo_cadena
def orfs_positivas(calculo_CG):
    return orf(calculo_CG)

def orfs_negativas(calculo_CG):
    secuencia_reverse = calculo_CG[::-1]
    return orf(secuencia_reverse)

#y ya podemos tener todos los orfs
total_positivo = orfs_positivas(calculo_CG)
total_negativo = orfs_negativas(calculo_CG)


#metemos todos los orfs juntos para asi tener mas facil el ordenar de mayor longitud a menor longitud y asi elegir los 6 primeros
todos_orfs = []
for seq in total_positivo.values():
    todos_orfs.append({'seq': seq, 'strand': '+'})
for seq in total_negativo.values():
    todos_orfs.append({'seq': seq, 'strand': '-'})

print('TODOS LOS ORFs:')
for i, orf_data in enumerate(todos_orfs, 1):
    print(f"{i}. Longitud: {len(orf_data['seq'])}, Strand: {orf_data['strand']}, Secuencia: {orf_data['seq']}")

todos_orfs.sort(key=lambda x: len(x['seq']), reverse=True)
mejores_orfs = todos_orfs[:6]
print('Los 6 más largos:')
for i, orf in enumerate(mejores_orfs, 1):
    print(f"{i}. Longitud: {len(orf['seq'])}, Strand: {orf['strand']}, Secuencia: {orf['seq']}")

#ahora elijo el más largo
mejor_orf= mejores_orfs[0]
print('El ORF más largo:', mejor_orf)

#a partir del mas largo generamos la proteina
#por cada elemento en un rango de 3 nucleotidos, sabiendo que un codon es tres nucleotidos de la secuencia de arn,
#que sustituya cada aa por cada codon, pero si el codon es uno de STOP, que acabe la proteina.
def traducir_proteina(mejor_orf, tabla_aa):
    arn = mejor_orf.replace('T', 'U')
    proteina= ""
    for i in range(0, len(secuencia_arn), 3):
        codon = secuencia_arn[i:i+3]
        if len(codon) == 3:
            aa = tabla_aa.get(codon, "")
            if aa == "*":
                break
            proteina += aa
    return proteina
print('La proteína resultante es:', proteina)

#para calcular el porcentaje de aa aromáticos, los cuales son Tirosina (Y), Fenilalanina (F), Triptófano (W)
aa_aromaticos = proteina.count('Y') + proteina.count('F') + proteina.count('W')
porcentaje = (aa_aromaticos / len(proteina))*100 

#para calcular la longitud relativa
lon_relativa = (len(mejor_orf['seq']) / len(calculo_CG))* 100

print('Porcentaje aromáticos:', porcentaje, '%')
print('Longitud relativa:', lon_relativa, '%')

```
