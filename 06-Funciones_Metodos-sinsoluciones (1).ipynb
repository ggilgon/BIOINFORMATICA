Las funciones se denominan con **def nombre_función(arg1,arg2):**

```python
# Como el primero es impar, sale del bucle y ya no mira nada más
#como el uno es el primer elemento y es impar, saldra como false aunque haya un numero par.
es_par([1,2,3])
```

```python
# Esta sería una solución correcta
# para que el resultado sea correcto, quitamos de ese bloque el return false, para que haga el bloque entero y una vez que hayan pasado todos los numero, ya ahi indicaria false en el caso de que todos sean impares
def es_par(lista_numeros):
    for numero in lista_numeros:
            if numero % 2 == 0:
                # si esta condición se cumple con return nos salimos de la lista
                return True
            else:
                pass
    # Cuando acaba de comprobar TODA la lista ahí sale del bloque
    return False
```

```python
# Ahora vamos a por un ejemplo un poco más complicado. Queremos que nos devuelva TODOS los numeros pares en un lista. 
# Si no hay números pares, nos va a devolver la lista vacía
def es_par(lista_numeros):
    # Creamos una lista vacía (lista de resultado)
    lista_pares = []
    # Parseamos cada elemento de nuestro input:
    for numero in lista_numeros:
        if numero % 2 == 0:
            # Si esto es verdad, añadimos nuestro elemento a la lista vacia que habiamos creado (en este caso todos los numero pares de la lista inicial en la lista de pares
            lista_pares.append(numero)
        else:
            pass
    return lista_pares
```

```python
# Vamos a crear ahora un pequeño juego. El jugador debe adivinar dónde está la letra "O"
# Empezamos creando una lista con tres elementos dónde la "O" se pone de forma aleatoria
#para cambiar aleatoriamente la posicion de un elemento de la lista necesitamos si o si utilizar random y shuffle para mezclar

from random import shuffle
mi_lista = [' ','O',' ']
def desordenar_lista(lista):
    shuffle(lista)
    return lista

desordenar_lista(mi_lista)
mi_lista
```

```python
# Ahora el jugador tiene que adivinar dónde está. Para ello hacemos una función
# while lo que ponga el jugador no es ni 0,1,2 sigue preguntando
#input es para que la persona pueda poner un texto 
def adivinacion():
    intento = ''
    while intento not in ['0','1','2']:
        intento = input("Esocoje un número: 0, 1, 2: ")
    return int(intento)
```

```python
adivinacion()
```

```python
# Ahora comprobamos el intento
#añadimos a la funcion los dos argumentos, en este caso la lista y el intento del jugador
#lista[intento] es para saber si coincide el numero con la posicion del o
def comprobacion(lista,intento):
    if lista[intento] == 'O':
        print('¡Has acertado!')
    else:
        print('¡Has fallado! Suerte para la próxima')
        print(lista)
```

```python
# Ahora vamos a juntarlo todo
#realizamos las funciones por separado, y tras ello lo juntamos para hacer la accion completa
mi_lista = [' ','O',' ']
lista_desordenada=desordenar_lista(mi_lista)
intento = adivinacion()
comprobacion(lista_desordenada,intento)
```

#### 1.- El menor de dos pares. Escribe una función que devuelva el menor de dos números, SOLAMENTE si los dos números son pares. En caso de que uno de los dos, o los dos, sean impares, devuelve el número más alto

```python
def menor_pares(lista_numeros):
    for numero in lista_numeros:
        if numero % 2 == 0:
            return min(lista_numeros)
        else:
            return max(lista_numeros)
```

```python
menor_pares([1,3])
```

```python
def menor_pares(n1,n2):
    if n1 % 2 == 0 and n2 % 2 == 0:
        return min(n1,n2)
    else:
        return max(n1,n2)
```

#### 7.- Escribe una función que busque numeros primos. 

```python
def es_primo(num):
    if num < 2:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

def primos_hasta(n):
    primos = []
    for i in range(2, n + 1):
        if es_primo(i):
            primos.append(i)
    return primos
```

```python
print(primos_hasta(20))
```

#### 8.- Juego de espías. Escribe una función que encuentre la secuencia 007 en una lista

```python
def juego_de_espias(lista):
    codigo = [0, 0, 7]
    for num in lista:
        if num == codigo[0]:
            codigo.pop(0)
        if not codigo:
            return True
    return False
```

```python
print(juego_de_espias([1, 2, 4, 0, 0, 7, 5]))   
```

```python
print(juego_de_espias([1, 0, 2, 4, 0, 5, 7]))
```

```python
print(juego_de_espias([1, 7, 2, 0, 4, 0, 5])) 
```

#### 9.- Juego de BLACKJACK. Dados tres numeros enteros entre el 1 y el 11, si su suma es menor o igual que 21, devuelve su suma. Si su suma es mayor que 21 y hay un 11, resta 10 al total de la suma. Finalmente, si con el ajuste es mayor que 21, se devuelve BUST.
##### 5,6,7 =18
##### 9,9,9 = BUST
##### 9,9,11 = 19

```python
def juego_blackjack(a, b, c):
    total = a + b + c
    if total <= 21:
        return total
    elif 11 in [a, b, c] and total - 10 <= 21:
        return total - 10
    else:
        return "BUST"
```

```python
print(juego_blackjack(5, 6, 7)) 
```

```python
print(juego_blackjack(9, 9, 9))
```

```python
print(juego_blackjack(9, 9, 11))
```
