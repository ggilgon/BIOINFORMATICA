#### 7.- Con la secuencia y la lista siguiente, escribe un algoritmo que cuente el número de veces que aparece un elemento de la lista en la secuencia. Pista: hacer un diccionario para almacenar los resultados puede ser buena idea

```python
dna = "AATGATGAACGAC" 
dinucleotides = ['AA','AT','AG','AC', 
                 'TA','TT','TG','TC', 
                 'GA','GT','GG','GC', 
                 'CA','CT','CG','CT'] 
```

```python
#definimos una función, hacemos un diccionario donde guardar los resultados, e indicamos que a partir de la lista, ver que elementos de lla se repiten o se encuentran en la secuencia de adn.
#por cada elemento en la secuencia, quiero que me extraiga el elemento en sí que se repite junto a su numero de veces de repetición.
def numero_veces(dinucleotides):
    resultados = {}
    for i in dinucleotides:
        if i in dna:
            resultados[i] = dna.count(i)
            #resultados.append(dna.count(i))
    return resultados
```

```python
resultados = numero_veces(dinucleotides)
```
