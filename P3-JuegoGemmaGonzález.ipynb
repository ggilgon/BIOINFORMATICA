#### Las reglas son las siguientes:

```python
# hacemos una funcion para el juego
# intentos para llevar un contador de las veces que lo ha intentado, e intento_ultimo para guardar el numero del último intento
def adivinar():
    #Explicamos de que trata el juego
    print('Bienvenidos a un juego muy divertido. Lo único que tienes que hacer es intentar adivinar el número que ha elegido el ordenador entre el 1 al 100. Si lo consigues a la primera, FELICIDADES, pero si no lo aciertas no pasa nada, tendras más intentos, además de pistas de si estas cerca o no jejeje. ¿Estás list@?')

    aleatorio = random.randint(1,100)
    intentos = 0
    intento_ultimo = None
    
    #hacemos el bucle del juego
    while True:
        inicio = input('Te toca, elige un número del 1 al 100')

        if inicio.isdigit() == False:
            print('Te has equivocado, eso no es un numero, intentalo de nuevo')
            continue
        else:
        
            numero = int(inicio)
        #verificar que esta dentro del rango
            if int(numero) <1 or int(numero) >100:
                print('FUERA DE LÍMITES')
                continue
    
    #contamos los intentos
            intentos = intentos + 1
    
    #si ha acertado el numero
            if numero == aleatorio:
                print('CORRECTO, has hecho', intentos, 'intentos')
                break

    #vemos si el numero que ha puesto esta cerca del que ha elegido el ordenador
    #y luego vemos si la distancia de antes es mayor o menor que la distancia entre el numero elegido del ordenador con el que ha elegido el jugador, pero en valor absoluto, ya que si hay un signo menos provoca errores.
            if intento_ultimo is None:
                if abs(aleatorio - intentos) <= 10:
                    print('CALIENTE!!')
                else:
                    print('FRIO FRIO')
            else:
                diferencia_intentoanterior = abs(aleatorio - intento_ultimo)
                diferencia_intentonuevo = abs(aleatorio - numero)

                if diferencia_intentonuevo < diferencia_intentoanterior:
                    print('MAS CALIENTE AAUUNNN')
                else:
                    print('MAS FRESQUITO :(')

        #para guardar el ultimo intento que ha hecho
            intento_ultimo = numero
    
```
